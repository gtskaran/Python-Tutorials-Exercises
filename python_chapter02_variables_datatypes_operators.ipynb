## 2. Multi-line Statements & Indentation

```python
# Explicit line continuation using \
a = 10 + 12 + 35 + \
    40 + 50 + 60 + \
    70 + 80 + 19
print('a (explicit continuation):', a)

# Implied line continuation inside parentheses, brackets, or braces
a = (10 + 12 + 35 +
     40 + 50 + 60 +
     70 + 80 + 19)
print('a (implicit continuation):', a)

colors = ['silver',
          'yellow',
          'red']
print('colors:', colors)

# Multiple statements in a single line using semicolons
a = 10; b = 20; c = 30
print('Multiple statements:', a, b, c)

# Python Indentation
for i in range(1, 10):
    print(i)
    if i == 6:
        break
```

## 4. Basic Data Types & String Slicing

```python
str1 = 'Python'
str2 = 'is an easy language!'

print(str1[0])       # Output: 'P'
print(str1[2:5])     # Output: 'tho'
print(str2[:2])      # Output: 'is'
print(str2[3:])      # Output: 'an easy language!'
print(str2[-1])      # Output: '!'
print(str1 + str2)   # Concatenation
print(str1 + ' ' + str2)
print(str1 * 2)      # Repetition
print('.' * 20)

# Checking types
a = 5
print(a, "is of type", type(a))
a = 5.0
print(a, "is of type", type(a))
a = 1 + 2j
print(a, "is complex number?", isinstance(1 + 2j, complex))
```

## 5. Python Collections: List, Tuple, Set, Dictionary

```python
# Python List (Mutable)
a = [5, 10, 15, 20, 25, 30, 35, 40]
print("a[2] =", a[2])
print("a[0:3] =", a[0:3])
print("a[5:] =", a[5:])
a[2] = 100
print("Updated List:", a)

# Python Tuple (Immutable)
t = (5, 'program', 1 + 3j)
print("t[0] =", t[0])
print("t[0:3] =", t[0:3])

# Python Set (Unordered, Unique elements)
s = {5, 2, 3, 1, 4}
print("Set:", s)
s_dup = {1, 2, 2, 3, 3, 3}
print("Set removes duplicates:", s_dup)

# Python Dictionary (Key-Value Pairs)
d = {1: 'ONE', 'key': 2}
print("d[1] =", d[1])
print("d['key'] =", d['key'])
```

## 7. Python Input, Output & Formatting

```python
# print() parameters
print(1, 2, 3, 4)
print(1, 2, 3, 4, sep='*')
print(1, 3, 5, 7, sep='#', end='&\n')

# String formatting
x, y = 5, 10
print('The value of x is {} and y is {}'.format(x, y))
print('I would like eat {0} and {1}'.format('Pizza', 'Burger'))
print('I had visited {1} and {0}'.format('Delhi', 'Mumbai'))
print('Hello {name}, {greeting}'.format(greeting='Good morning', name='Ramesh'))

# Old style % formatting
val = 12.3456789
print('The value of x is %3.2f' % val)
print('The value of x is %3.4f' % val)
```

## 8. Python Operators

```python
# Arithmetic Operators
x, y = 15, 4
print('x + y =', x + y)
print('x - y =', x - y)
print('x * y =', x * y)
print('x / y =', x / y)
print('x // y =', x // y)
print('x ** y =', x ** y)

# Comparison Operators
x, y = 10, 12
print('x > y is', x > y)
print('x < y is', x < y)
print('x == y is', x == y)
print('x != y is', x != y)

# Logical Operators
x, y = True, False
print('x and y is', x and y)
print('x or y is', x or y)
print('not x is', not x)

# Bitwise Operators
x, y = 10, 4
print('x & y =', x & y)
print('x | y =', x | y)
print('~x =', ~x)
print('x ^ y =', x ^ y)
print('x >> 2 =', x >> 2)
print('x << 2 =', x << 2)

# Identity & Membership Operators
x1, y1 = 5, 5
x2, y2 = 'Hello', 'Hello'
x3, y3 = [1, 2, 3], [1, 2, 3]
print('Identity (x1 is not y1):', x1 is not y1)
print('Identity (x3 is y3):', x3 is y3)

str_val = 'Hello world'
dict_val = {1: 'a', 2: 'b'}
print('Membership ("H" in str_val):', 'H' in str_val)
print('Membership (1 in dict_val):', 1 in dict_val)
```

### Bitwise Operators in Python

| Operator | Name              | Example | Result (assuming a = 5, b = 3) |
|----------|-------------------|---------|--------------------------------|
| `&`      | AND               | a & b   | 1 (0101 & 0011 = 0001)         |
| &vert; | OR                | a &#124; b | 7 (0101 &#124; 0011 = 0111)   |
| `^`      | XOR               | a ^ b   | 6 (0101 ^ 0011 = 0110)         |
| `~`      | NOT (bitwise neg) | ~a      | -6 (in 2’s complement)         |
| `<<`     | Left Shift        | a << 1  | 10 (0101 << 1 = 1010)          |
| `>>`     | Right Shift       | a >> 1  | 2 (0101 >> 1 = 0010)           |

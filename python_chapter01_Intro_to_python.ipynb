# Chapter 1: Introduction to Python

Python is a cross-platform, general-purpose, interpreted, interactive, object-oriented, and high-level programming language. It is free and open-source, designed with a focus on code readability and clean syntax.

## 1. History of Python

* **Creator:** Guido van Rossum
* **Timeline:** Developed during 1985–1990 (late 1980s and early 1990s) at the National Research Institute for Mathematics and Computer Science (CWI) in the Netherlands.
* **Influences:** Derived from languages such as ABC, Modula-3, C, C++, Algol-68, SmallTalk, Unix shell, and other scripting languages.
* **Licensing & Maintenance:** Copyrighted under the GNU General Public License (GPL) and maintained by a core development team.

## 2. Core Features of Python

* **Easy-to-learn & Read:** Simple structure, clearly defined syntax, and frequent use of English keywords.
* **Easy-to-maintain:** Source code is straightforward and easy to maintain.
* **Broad Standard Library:** Portable and cross-platform compatible across Windows, macOS, Linux, and Unix.
* **Interactive Mode:** Allows direct testing and debugging of code snippets.
* **Portable & Extendable:** Runs on various hardware platforms; low-level modules can be added to the interpreter.
* **Database Support & GUI:** Interfaces to major commercial databases and supports GUI frameworks.
* **Automatic Garbage Collection:** Handles memory management under the hood.

## 3. Starting the Interpreter

### Execution Modes:
1. **Immediate Mode:** Launch the interpreter prompt (`>>>`) to execute statements directly.
2. **Script Mode:** Save code in a text file with a `.py` extension (e.g., `PythonWorld.py`) and run it via terminal (`python PythonWorld.py`).

```python
# Immediate Mode Simulation
print("Hello World from Script Mode!")
result = 1 + 1
print("1 + 1 =", result)
```

## 4. Identifiers and Keywords

### Rules for Valid Identifiers:
1. The first character must be a letter (`a-z`, `A-Z`) or an underscore (`_`).
2. Subsequent characters can be alphanumeric (`a-z`, `A-Z`, `0-9`) or underscores.
3. Identifiers are **case-sensitive**.
4. Identifiers cannot start with numbers, and no special symbols other than underscores are allowed.

### Reserved Keywords:
Keywords are reserved strings defined by Python syntax that cannot be used as variable identifiers.

```python
import keyword

# Programmatically retrieve reserved keywords
print("Total Python Keywords:", len(keyword.kwlist))
print(keyword.kwlist)
```

## 5. Memory Management

* Variables do not require explicit type declarations ahead of time.
* Variable types are inferred dynamically upon assignment.
* Automatic garbage collection handles de-allocation.
* Variable names can be reused ("recycled").
* The `del` statement allows explicit reference removal.

```python
# Memory management and dynamic typing
data = 100
print("Initial value:", data)

# Recycling variable name
data = "Now a String"
print("Recycled variable:", data)

# Explicit de-allocation
del data
```
